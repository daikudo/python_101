```python
import requests
```

## поиск по API

```python
params = {"q": "Van Gogh", "hasImages": "true"}
base_api_url = "https://collectionapi.metmuseum.org/public/collection/v1/search"
search = requests.get (base_api_url, params=params).json()
```

```python
search
```

```python
for key in search:
    print (key)
```

```python
object_ids = search["objectIDs"]
```

```python
first_object_ids = object_ids[0]
```

```python
first_object_ids
```

```python
base_object_url = "https://collectionapi.metmuseum.org/public/collection/v1/objects/"
```

```python
first_object_url = base_object_url + str(first_object_ids)
```

```python
our_path = "output/met_Van_Gogh/"
```

```python
for obj_id in object_ids[21:26]:
    obj_url = base_object_url + str(obj_id)
    obj_response = requests.get(obj_url).json()
    primary_img_url = obj_response["primaryImage"]
    artistName = obj_response["artistDisplayName"]
    title = obj_response["title"]

    extension = primary_img_url.split(".")[-1]

    filename = f"{title}_{artistName}.{extension}"
    filename = filename.replace(" ", "_")

    full_path = our_path + filename

    if primary_img_url:
        img_response = requests.get(primary_img_url)
        status_code = img_response.status_code
        if status_code == 200:
            with open(full_path, "wb") as f:
                f.write(img_response.content)
            print(f"Успешно сохранили: {filename}")

    print(filename)
    print(primary_img_url)
```
